```python
import pandas as pd
import matplotlib.pyplot as plt

# read dataset
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/climate_change_data.csv')

# average temperature per country
avg_temp = df.groupby('Country')['Temperature'].mean()

# create chart
avg_temp.plot(kind='bar', figsize=(12,6))

# labels and title
plt.xlabel('Country')
plt.ylabel('Average Temperature')
plt.title('Average Temperature per Country')

# rotate country names
plt.xticks(rotation=90)

# show chart
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# read dataset
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/climate_change_data.csv')

# average temperature per country
avg_temp = df.groupby('Country')['Temperature'].mean()

# top 5 coldest countries
top5_coldest = avg_temp.sort_values().head(5)

# create bar chart
top5_coldest.plot(kind='bar')

# labels and title
plt.xlabel('Country')
plt.ylabel('Average Temperature')
plt.title('Top 5 Coldest Countries')

# show chart
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# read dataset
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/climate_change_data.csv')

# average temperature per country
avg_temp = df.groupby('Country')['Temperature'].mean()

# top 5 hottest countries
top5 = avg_temp.sort_values(ascending=False).head(5)

# create bar chart
top5.plot(kind='bar')

# labels and title
plt.xlabel('Country')
plt.ylabel('Average Temperature')
plt.title('Top 5 Hottest Countries')

# show chart
plt.show()
```
